```python
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib_scalebar.scalebar import ScaleBar

# Define the path to your shapefile
file_path = r"C:\Users\daves\OneDrive\Pessoal\Artigo\Economic History Review\Code\MT_RG_Imediatas_2024\MT_RG_Imediatas_2024.shp"

# Load the shapefile and project it to a CRS with meters as units
gdf = gpd.read_file(file_path)
gdf = gdf.to_crs(epsg=32721) # UTM zone 21S, units are meters

# --- Plotting Setup ---
fig, ax = plt.subplots(1, 1, figsize=(12, 14))
ax.set_aspect('equal') # Ensure the map has the correct proportions

# Define a list of hatch patterns to cycle through
hatch_patterns = ['/', '\\', '|', '-', '+', 'x', 'o', 'O', '.', '*']
unique_regions = gdf['NM_RGI'].unique()
legend_handles = []

# Loop through each region to apply a unique hatch pattern
for i, region in enumerate(unique_regions):
    region_gdf = gdf[gdf['NM_RGI'] == region]
    hatch = hatch_patterns[i % len(hatch_patterns)]
    
    region_gdf.plot(ax=ax,
                    facecolor='lightgray',
                    edgecolor='black',
                    hatch=hatch)
    
    legend_handles.append(mpatches.Patch(facecolor='lightgray',
                                         edgecolor='black',
                                         hatch=hatch,
                                         label=region))

# Create the legend from our custom handles
ax.legend(handles=legend_handles,
          title="Immediate Regions",
          loc='upper center',
          bbox_to_anchor=(0.5, -0.01), # Position legend below the plot
          ncol=3,
          fancybox=True,
          shadow=True)

# --- Add Custom Dual Scale Bars ---

# Create a custom kilometers scale bar
def add_kilometers_scale_bar(ax, location="lower right"):
    """Add a custom kilometers scale bar"""
    
    # Get the current axis limits to determine appropriate scale
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    # Use 200 km as specified
    scale_length_km = 200
    scale_length_m = scale_length_km * 1000  # Convert to meters
    
    # Position the scale bar
    if location == "lower right":
        x_pos = xlim[1] - (xlim[1] - xlim[0]) * 0.25
        y_pos = ylim[0] + (ylim[1] - ylim[0]) * 0.05
    else:
        x_pos = xlim[0] + (xlim[1] - xlim[0]) * 0.05
        y_pos = ylim[0] + (ylim[1] - ylim[0]) * 0.05
    
    # Draw the scale bar line
    ax.plot([x_pos, x_pos + scale_length_m], [y_pos, y_pos], 
            color='black', linewidth=2, solid_capstyle='butt')
    
    # Add tick marks at the ends
    tick_height = (ylim[1] - ylim[0]) * 0.01
    ax.plot([x_pos, x_pos], [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=2)
    ax.plot([x_pos + scale_length_m, x_pos + scale_length_m], 
            [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=2)
    
    # Add labels (removed "km" from the main label)
    ax.text(x_pos + scale_length_m/2, y_pos + tick_height*2, 
            'Kilometers', 
            ha='center', va='bottom', fontsize=10, color='black')
    ax.text(x_pos + scale_length_m/2, y_pos + tick_height*4, 
            f'{scale_length_km}', 
            ha='center', va='bottom', fontsize=10, color='black')
    ax.text(x_pos, y_pos - tick_height*2, '0', 
            ha='center', va='top', fontsize=9, color='black')
    ax.text(x_pos + scale_length_m, y_pos - tick_height*2, 
            f'{scale_length_km}', 
            ha='center', va='top', fontsize=9, color='black')

# Add a custom scale bar for miles
def add_miles_scale_bar(ax, location="lower left"):
    """Add a custom miles scale bar"""
    
    # Get the current axis limits to determine appropriate scale
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    # Use 200 km as the base (same as kilometers scale bar)
    scale_length_km = 200
    scale_length_m = scale_length_km * 1000  # Convert to meters
    
    # Convert to miles with 2 decimal places
    scale_length_miles = scale_length_km * 0.621371
    
    # Position the scale bar
    if location == "lower left":
        x_pos = xlim[0] + (xlim[1] - xlim[0]) * 0.05
        y_pos = ylim[0] + (ylim[1] - ylim[0]) * 0.05
    elif location == "lower right":
        x_pos = xlim[1] - (xlim[1] - xlim[0]) * 0.25
        y_pos = ylim[0] + (ylim[1] - ylim[0]) * 0.05
    else:
        x_pos = xlim[0] + (xlim[1] - xlim[0]) * 0.05
        y_pos = ylim[0] + (ylim[1] - ylim[0]) * 0.05
    
    # Draw the scale bar line
    ax.plot([x_pos, x_pos + scale_length_m], [y_pos, y_pos], 
            color='black', linewidth=2, solid_capstyle='butt')
    
    # Add tick marks at the ends
    tick_height = (ylim[1] - ylim[0]) * 0.01
    ax.plot([x_pos, x_pos], [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=2)
    ax.plot([x_pos + scale_length_m, x_pos + scale_length_m], 
            [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=2)
    
    # Add labels with proper decimal formatting
    ax.text(x_pos + scale_length_m/2, y_pos + tick_height*2, 
            f'{scale_length_miles:.2f} Miles', 
            ha='center', va='bottom', fontsize=10, color='black')
    ax.text(x_pos, y_pos - tick_height*2, '0', 
            ha='center', va='top', fontsize=9, color='black')
    ax.text(x_pos + scale_length_m, y_pos - tick_height*2, 
            f'{scale_length_miles:.2f}', 
            ha='center', va='top', fontsize=9, color='black')

# Add the custom scale bars to the plot
add_kilometers_scale_bar(ax, location="lower right")
add_miles_scale_bar(ax, location="lower left")

# Turn off the axes and frame for a clean map
ax.axis('off')

# Adjust layout to prevent the legend from being cut off
# The 'rect' parameter creates space at the bottom for the legend
plt.tight_layout(rect=[0, 0.05, 1, 1]) 

# Display the plot
plt.show()
```

```python
# --- 1. DEFINE FILE PATHS ---
brazil_path = r"C:\Users\daves\OneDrive\Pessoal\Artigo\Economic History Review\Code\BR_UF_2024\BR_UF_2024.shp"
mt_regions_path = r"C:\Users\daves\OneDrive\Pessoal\Artigo\Economic History Review\Code\MT_RG_Imediatas_2024\MT_RG_Imediatas_2024.shp"

# --- 2. LOAD AND PREPARE DATA ---
gdf_brazil = gpd.read_file(brazil_path)
gdf_mt_regions = gpd.read_file(mt_regions_path)

# Convert to EPSG:5880 (Polyconic) -> Meters
gdf_brazil = gdf_brazil.to_crs(epsg=5880)
gdf_mt_regions = gdf_mt_regions.to_crs(epsg=5880)

gdf_mt_state = gdf_brazil[gdf_brazil['SIGLA_UF'] == 'MT']

# --- 3. SETUP THE PLOT ---
fig = plt.figure(figsize=(18, 12))
fig.patch.set_facecolor('white')

# ax1: Brazil Map (Inset)
ax1 = fig.add_axes([0.05, 0.3, 0.1, 0.6]) 
# ax2: Mato Grosso Map (Main)
ax2 = fig.add_axes([0.20, 0.3, 0.6, 0.6])

# --- 4. PLOT THE MAPS ---
gdf_brazil.plot(ax=ax1, color='lightgray', edgecolor='white')
gdf_mt_state.plot(ax=ax1, color='#333333', edgecolor='white')

hatch_patterns = ['/', '\\', '|', '-', '+', 'x', 'o', 'O', '.', '*']
unique_regions = gdf_mt_regions['NM_RGI'].unique()
legend_handles = []

for i, region in enumerate(unique_regions):
    region_gdf = gdf_mt_regions[gdf_mt_regions['NM_RGI'] == region]
    hatch = hatch_patterns[i % len(hatch_patterns)]
    region_gdf.plot(ax=ax2,
                    facecolor='lightgray',
                    edgecolor='black',
                    hatch=hatch)
    legend_handles.append(mpatches.Patch(facecolor='lightgray',
                                         edgecolor='black',
                                         hatch=hatch,
                                         label=region))

# --- 5. DEFINE FLEXIBLE SCALE BAR FUNCTIONS ---

def add_kilometers_scale_bar(ax, scale_length_km=200, location="lower right", 
                             fontsize=12, linewidth=3, y_offset_factor=0.05):
    """
    Adds a custom kilometers scale bar.
    - scale_length_km: Distance in km to represent
    - y_offset_factor: Fine-tune vertical placement (percent of axis height)
    """
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    scale_length_m = scale_length_km * 1000  # Convert to meters
    
    # Calculate positions
    width = xlim[1] - xlim[0]
    height = ylim[1] - ylim[0]
    
    if location == "lower right":
        x_pos = xlim[1] - width * 0.25
        y_pos = ylim[0] + height * y_offset_factor
    elif location == "lower left":
        x_pos = xlim[0] + width * 0.05
        y_pos = ylim[0] + height * y_offset_factor
    elif location == "below center": # Special case for Brazil map
        x_pos = xlim[0] + width * 0.2
        y_pos = ylim[0] - height * 0.05 # Push below the map
    
    # Draw Line
    ax.plot([x_pos, x_pos + scale_length_m], [y_pos, y_pos], 
            color='black', linewidth=linewidth, solid_capstyle='butt', clip_on=False)
    
    # Draw Ticks
    tick_height = height * 0.015
    # Adjust tick height for 'below center' since it's outside the axis
    if location == "below center": tick_height = height * 0.03
        
    ax.plot([x_pos, x_pos], [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=linewidth, clip_on=False)
    ax.plot([x_pos + scale_length_m, x_pos + scale_length_m], 
            [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=linewidth, clip_on=False)
    
    # Labels
    ax.text(x_pos + scale_length_m/2, y_pos + tick_height, 
            f'{scale_length_km} km', 
            ha='center', va='bottom', fontsize=fontsize, weight='bold', color='black')
    
    ax.text(x_pos, y_pos - tick_height*1.5, '0', 
            ha='center', va='top', fontsize=fontsize, color='black')
    ax.text(x_pos + scale_length_m, y_pos - tick_height*1.5, 
            f'{scale_length_km}', 
            ha='center', va='top', fontsize=fontsize, color='black')

def add_miles_scale_bar(ax, scale_length_km=200, location="lower left", 
                        fontsize=12, linewidth=3, y_offset_factor=0.05):
    """
    Adds a custom miles scale bar based on a km reference length.
    """
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    scale_length_m = scale_length_km * 1000
    scale_length_miles = scale_length_km * 0.621371
    
    width = xlim[1] - xlim[0]
    height = ylim[1] - ylim[0]
    
    if location == "lower left":
        x_pos = xlim[0] + width * 0.05
        y_pos = ylim[0] + height * y_offset_factor
    elif location == "lower right":
        x_pos = xlim[1] - width * 0.25
        y_pos = ylim[0] + height * y_offset_factor
    elif location == "below center":
        x_pos = xlim[0] + width * 0.2
        # Stack below the KM bar (which is at -0.05)
        y_pos = ylim[0] - height * 0.15 
        
    # Draw Line
    ax.plot([x_pos, x_pos + scale_length_m], [y_pos, y_pos], 
            color='black', linewidth=linewidth, solid_capstyle='butt', clip_on=False)
    
    # Draw Ticks
    tick_height = height * 0.015
    if location == "below center": tick_height = height * 0.03
        
    ax.plot([x_pos, x_pos], [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=linewidth, clip_on=False)
    ax.plot([x_pos + scale_length_m, x_pos + scale_length_m], 
            [y_pos - tick_height/2, y_pos + tick_height/2], 
            color='black', linewidth=linewidth, clip_on=False)
    
    # Labels
    ax.text(x_pos + scale_length_m/2, y_pos + tick_height, 
            f'{scale_length_miles:.2f} Miles', 
            ha='center', va='bottom', fontsize=fontsize, weight='bold', color='black')
    
    ax.text(x_pos, y_pos - tick_height*1.5, '0', 
            ha='center', va='top', fontsize=fontsize, color='black')
    ax.text(x_pos + scale_length_m, y_pos - tick_height*1.5, 
            f'{scale_length_miles:.2f}', 
            ha='center', va='top', fontsize=fontsize, color='black')


# --- 6. APPLY SCALE BARS ---

# A. Mato Grosso Map (Main) - 200 km / 124.27 Miles
add_kilometers_scale_bar(ax2, scale_length_km=200, location="lower right")
add_miles_scale_bar(ax2, scale_length_km=200, location="lower left")

# B. Brazil Map (Inset) - 1000 km / 621.37 Miles
# We use a larger scale (1000km) because the map is zoomed out
# We also reduce font size/linewidth slightly to fit the smaller inset area
add_kilometers_scale_bar(ax1, scale_length_km=1000, location="below center", 
                         fontsize=9, linewidth=2)

add_miles_scale_bar(ax1, scale_length_km=1000, location="below center", 
                    fontsize=9, linewidth=2)


# --- 7. PLACE LEGEND ---
fig.legend(handles=legend_handles,
           title="Immediate Regions",
           loc='lower center',
           bbox_to_anchor=(0.5, 0.1),
           ncol=5,
           fancybox=True,
           shadow=True,
           fontsize='large',
           title_fontsize='x-large')

# --- 8. DRAW ARROW ---
start_point = (gdf_mt_state.geometry.centroid.iloc[0].x, gdf_mt_state.geometry.centroid.iloc[0].y)

ax1.annotate("",
             xy=(0, 0.5),
             xycoords=ax2.transAxes,
             xytext=start_point,
             textcoords=ax1.transData,
             arrowprops=dict(arrowstyle="->",
                             connectionstyle="arc3,rad=0.2",
                             color="black",
                             lw=2.2))

# --- 9. FINAL TOUCHES AND SAVE ---
ax1.axis('off')
ax2.axis('off')

plt.savefig('inset_map_brazil_mt_final.jpg', dpi=1200, bbox_inches='tight')
print("Map saved as 'inset_map_brazil_mt_final.jpg'.")
```
